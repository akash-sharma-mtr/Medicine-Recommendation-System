```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
```

```python
X = dataset.drop('prognosis', axis=1)
y = dataset['prognosis']
```

```python
le = LabelEncoder()
le.fit(y)
Y = le.transform(y)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, Y, test_size=0.25, random_state=42)
```

```python
from sklearn.datasets import make_classification
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import MultinomialNB

from sklearn.metrics import accuracy_score, confusion_matrix
import numpy as np
```

```python
models = {
    'Random Forest': RandomForestClassifier(n_estimators=100, random_state=42),
    'Gradient Boosting': GradientBoostingClassifier(n_estimators=100, random_state=42),
    'SVC': SVC(kernel='linear', random_state=42),
    'KNN': KNeighborsClassifier(n_neighbors=5),
    'Naive Bayes': MultinomialNB()
}

for model_name, model in models.items():
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    accuracy = accuracy_score(y_test, y_pred)
    cm = confusion_matrix(y_test, y_pred)
    
    print(f"{model_name} Accuracy: {accuracy:.2f}")
    print(f"{model_name} Confusion Matrix:\n{cm}\n")
    print(np.array2string(cm, separator=', '))
```

```python
svc = SVC(kernel='linear', random_state=42)
svc.fit(X_train, y_train)
y_pred = svc.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)
```

```python
import pickle
pickle.dump(svc, open('Models/model.pkl', 'wb'))
```

```python
svc = pickle.load(open('Models/model.pkl', 'rb'))
print("Actual Label : ", svc.predict(X_test.iloc[0].values.reshape(1, -1)))
print("Predicted Label : ", y_test[0])
```

```python
def get_prediction(symtoms) :
    input_vector = np.zeros(len(symtoms_dict))
    
    for item in symtoms:
        input_vector[symtoms_dict[item]] = 1
    
    return disease_dict[svc.predict([input_vector])[0]]
```

```python
symptoms = input("Enter Your Symptoms : ").lower()
patient_symptoms = [s.strip() for s in symptoms.split(',')]

patient_symptoms = [sym.strip() for sym in patient_symptoms]
patient_symptoms
```

```python
predicted_disease = get_prediction(patient_symptoms)
print(f"Predicted Disease: {predicted_disease}")
```

```python
def helper(disease) :
    
    desc = description[description["Disease"] == disease]["Description"]
    
    desc = " ".join([i for i in desc])
    
    precaution = precautions[precautions["Disease"] == disease][
        ["Precaution_1" , "Precaution_2" , "Precaution_3" , "Precaution_4"]
    ]
    precaution = [i for i in precaution.values]
    
    symptoms_ = sys_des[sys_des["Disease"] == disease][
        ["Symptom_1" , "Symptom_2" , "Symptom_3" , "Symptom_4"]
    ]
    symptoms_list = [i for i in symptoms_.values]
    
    medicine = medications[medications["Disease"] == disease]["Medication"]
    medicine = [i for i in medicine.values]
  
    diet = diets[diets["Disease"] == disease]["Diet"]
    diet = [i for i in diet.values]
  
    workout = workout_data[workout_data["disease"] == disease]["workout"]
    workout = [i for i in workout.values]

    return desc , precaution , medicine , diet , workout , symptoms_list
```
